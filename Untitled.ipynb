```python
# Day 2
```

```python
df_train = pd.read_csv("application_train.csv")
df_test = pd.read_csv("application_test.csv")

print(df_train.shape)
print(df_test.shape)
```

```python
df_train["AGE"] = (-df_train['DAYS_BIRTH'] / 365).astype(int)
df_test["AGE"] = (-df_test["DAYS_BIRTH"] / 365).astype(int)

print(df_train["AGE"].describe())
print(df_test["AGE"].describe())
```

```python
numeric_cols = df_train.select_dtypes(include='number').columns.tolist()

stats = df_train[numeric_cols].describe().T
stats['range'] = stats['max'] - stats['min']
print(stats[['min', 'max', 'mean', 'range']].sort_values('max', ascending=False).head(40))
```

```python
days_cols = []
for col in df_train.columns:
    if "DAYS" in col:
        days_cols.append(col)

print(days_cols)
print()

for col in days_cols:
    max_val = df_train[col].max()
    pos_count = (df_train[col] > 0).sum()
    print(f"{col:<35} max={max_val:>10.0f}   positive values={pos_count}")
```

```python
df_train["DAYS_EMPLOYED"] = df_train["DAYS_EMPLOYED"].replace(365243, np.nan)
df_test['DAYS_EMPLOYED'] = df_test['DAYS_EMPLOYED'].replace(365243, np.nan)
```

```python
cat_cols = df_train.select_dtypes(include="object").columns.tolist()

endcoder = OrdinalEncoder(handle_unknown="use_encoded_value", unknown_value=-1)

df_train[cat_cols] = endcoder.fit_transform(df_train[cat_cols])
df_test[cat_cols] = endcoder.transform(df_test[cat_cols])
print("Encoding done!")
df_train[cat_cols].sample(10)
```

```python
x = df_train.drop(columns=["TARGET", "SK_ID_CURR"])
y = df_train["TARGET"]

x_test = df_test.drop(columns=['SK_ID_CURR'])

model = lgb.LGBMClassifier(n_estimators=500, learning_rate=0.01, random_state=42, n_jobs=-1)

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

scores = cross_val_score(model, x, y, cv=cv, scoring="roc_auc")

print("AUC per fold:", scores.round(4))
print()
print(f"Mean CV AUC : {scores.mean():.4f}")
print(f"Std         : {scores.std():.4f}")
print()
```

```python
model.fit(x, y)
test_preds = model.predict_proba(x_test)[:, 1]
```

```python
test_preds[:5].round(4)
```

```python
importance = pd.Series(model.feature_importances_, index=x.columns)
importance = importance.sort_values(ascending=False).head(20)

importance.plot(kind="barh", figsize=(8, 8), color="steelblue")
plt.xlabel("Importance Score")
plt.title("Top 20 Most Important Features")
plt.gca().invert_yaxis()
plt.tight_layout()
plt.savefig('feature_importance.png', dpi=150)
plt.show()
print("=== Top 10 Features ===")
print()
print(importance.head(10).to_string())
```

```python
train_preds = model.predict_proba(x)[:, 1]
fpr_day2, tpr_day2, thresholds = roc_curve(y, train_preds)

train_auc = roc_auc_score(y, train_preds)

plt.figure(figsize=(8, 8))
plt.plot(fpr_day2, tpr_day2, color="steelblue", label=f"Our model AUC = {train_auc:.4f}")
plt.plot([0, 1], [0, 1], color='tomato', linestyle='--', label='Random model (AUC = 0.5)')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('ROC Curve - How well does our model catch defaulters?')
plt.legend()
plt.tight_layout()
plt.savefig('roc_curve.png', dpi=150)
plt.show()
```

```python
submission = pd.DataFrame({
    'SK_ID_CURR': df_test['SK_ID_CURR'],
    'TARGET': test_preds
})

submission.to_csv('submission.csv', index=False)
```

```python
inst_agg = installments.drop(columns=['SK_ID_PREV']).groupby('SK_ID_CURR').agg(['mean', 'max', 'min', 'sum'])

inst_agg.columns = ['INST_' + '_'.join(col) for col in inst_agg.columns]
inst_agg = inst_agg.reset_index()

print("Installments aggregated shape:", inst_agg.shape)
print("Sample columns:")
print(inst_agg.columns[:10].tolist())
```

```python
df_train = df_train.merge(inst_agg, on='SK_ID_CURR', how='left')
df_test = df_test.merge(inst_agg, on='SK_ID_CURR', how='left')

print("Train shape after adding installments:", df_train.shape)
print("Test shape after adding installments :", df_test.shape)
print("New features added:", inst_agg.shape[1] - 1)
```

```python
X = df_train.drop(columns=['TARGET', 'SK_ID_CURR'])
y = df_train['TARGET']
X_test = df_test.drop(columns=['SK_ID_CURR'])

model = lgb.LGBMClassifier(
    n_estimators=500,
    learning_rate=0.05,
    class_weight='balanced',
    random_state=42
)

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
scores = cross_val_score(model, X, y, cv=cv, scoring='roc_auc')

print("AUC per fold:", scores.round(4))
print()
print(f"Day 3 AUC : 0.7752")
print(f"Day 4 AUC : 0.7759")
print(f"Day 5 AUC : {scores.mean():.4f}")
print(f"Improvement over Day 3 : {scores.mean() - 0.7752:+.4f}")
```
